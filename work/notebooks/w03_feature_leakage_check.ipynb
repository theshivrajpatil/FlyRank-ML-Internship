## 3. The leakage hunt

*Attack your own features: label-derived columns, future windows, product flags. Show the test.*

```python

leakage = [
    "trend_pct",
    "trend_direction"
]

print("leakage columns:")
for col in leakage:
    if col in df.columns:
        print("-", col)
```

```python
print("""
Leakage Result:

While checking the features, I found that 'trend_pct' is closely related to the target value.
If I include this column in the model, it would indirectly reveal the answer before prediction.
This would make the model appear much more accurate than it actually is, which is not realistic.
To keep the model fair and reliable, I removed 'trend_pct' from the feature vector.
""")
```
